```python
# Atividade 1
a = int(input("Digite um número: "))
if a%2 == 0 :
  print("Seu número é par!")
else:
  print("Seu número é impar!")

# Atividade 2
n = int(input("Digite um número inteiro positivo: "))
fatorial = 1
for i in range(1, n + 1):
    fatorial *= i
print(f"O fatorial de {n} é {fatorial}.")

# Atividade 3
n = int(input("Digite um número inteiro positivo: "))
soma = 0
for i in range(1, n + 1):
    soma += i
print(f"A soma dos primeiros {n} números naturais é {soma}.")

# Atividade 4
b = int(input("Digite um número inteiro e positivo: "))
for i in range(1, 11):
    print(f"{b} x {i} = {b * i}")

# Atividade 5
n = int(input("Digite um número inteiro positivo: "))
for i in range (0 , n+1):
  print(n)
  n-=1


# Atividade 6
numeros = []
for i in range(5):
    numero = float(input(f"Digite o {i + 1}º número: "))
    numeros.append(numero)
maior = max(numeros)
menor = min(numeros)

print(f"\nO maior número digitado foi: {maior}")
print(f"O menor número digitado foi: {menor}")


# Atividade 7
nUm = float(input("Digite o primeiro número: "))
nDois = float(input("Digite o segundo número: "))
multiplicacao = nUm * nDois
divisao = nUm/nDois
adicao = nUm+nDois
subtracao = nUm-nDois
print(f'O resultado das operações é:\nAdição: {adicao} \nSubtraçao:{subtracao}\nDivisão: {divisao}\nMultiplicação: {multiplicacao}')


# Atividade 8
n = int(input("Digite um número inteiro: "))
if n <= 1:
    print(f"{n} não é primo.")
else:
    primo = True
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            primo = False
            break
    print(f"{n} é primo!" if primo else f"{n} não é primo.")


# Atividade 9
n = int(input("Digite quantos termos da sequência de Fibonacci deseja: "))
if n <= 0:
    print("Erro: Digite um número maior que 0!")
elif n == 1:
    print("Sequência de Fibonacci com 1 termo:\n[0]")
elif n == 2:
    print("Sequência de Fibonacci com 2 termos:\n[0, 1]")
else:
    fibonacci = [0, 1]
    for i in range(2, n):
        proximo_termo = fibonacci[-1] + fibonacci[-2]
        fibonacci.append(proximo_termo)
    print(f"Sequência de Fibonacci com {n} termos:\n{fibonacci}")


# Atividade 10
lista = [1,5,1,7,9,4,6]
lista.reverse()
print(lista)



# Atividade 11
palavra = input("Digite uma palavra: ").strip().lower()
if palavra == palavra[::-1]:
    print(f'"{palavra}" é um palíndromo!')
else:
    print(f'"{palavra}" não é um palíndromo.')


# Atividade 12
notaUm = float(input("Digite a primeira nota: "))
notaDois = float(input("Digite a segunda nota: "))
notaTres = float(input("Digite a terceira nota: "))
notaQuatro = float(input("Digite a quarta nota: "))
media = (notaUm + notaDois + notaTres + notaQuatro)/4
if media>= 7 :
  print("Aluno aprovado!")
else:
  print("Aluno reprovado!")

# Atividade 13
numeros = [5, 2, 9, 1, 7]
numeros_ordenados = sorted(numeros)
print("Ordem crescente:", numeros_ordenados)
print("Lista original:", numeros)
```
